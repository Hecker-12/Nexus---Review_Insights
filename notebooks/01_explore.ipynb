```python
import pandas as pd

def read_ft(path, n=None):
    rows = []
    with open(path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if n is not None and i >= n:
                break
            label, text = line.split(" ", 1)
            rows.append((label, text.strip()))
    return pd.DataFrame(rows, columns=["label", "text"])

df = read_ft("../data/raw/train.ft.txt", n=100_000)
df.head()
```

```python
print(df.shape)
print(df["label"].value_counts())

print(df["text"].str.split().str.len().describe())

for t in df.sample(15, random_state=1)["text"]:
    print("-", t, "\n")
```

```python
print("duplicates:", df.duplicated(subset="text").sum())
print("missing:", df["text"].isna().sum())
print("empty:", (df["text"].str.strip() == "").sum())
```
