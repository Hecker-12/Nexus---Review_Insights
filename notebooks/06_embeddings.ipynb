```python
import pandas as pd
from sentence_transformers import SentenceTransformer, util

df = pd.read_parquet("../data/processed/reviews_clean.parquet")
model = SentenceTransformer("all-MiniLM-L6-v2")

# 1. Tiny sanity check
sents = ["The battery dies very quickly",
         "Power doesn't last long at all",
         "The plot was brilliant and gripping"]
emb = model.encode(sents)
print(emb.shape)
print(util.cos_sim(emb, emb))
```

```python
import time
sample = df["text"].head(1000).tolist()
start = time.time()
e = model.encode(sample, batch_size=64, show_progress_bar=True)
print(f"{time.time() - start:.1f} s for 1000 reviews")
```

```python
import numpy as np

texts = df["text"].tolist()
embeddings = model.encode(texts, batch_size=64, show_progress_bar=True)
print(embeddings.shape)          # expect (50000, 384)

np.save("../data/processed/embeddings.npy", embeddings)
```
